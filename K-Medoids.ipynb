```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn_extra.cluster import KMedoids
from scipy.spatial.distance import cdist

# Sample dataset (2D points)
data = np.array([
    [2, 6], [3, 4], [4, 7], [6, 2], [7, 5], [8, 3],[10,10]
])

# Number of clusters
k = 2

# Applying K-Medoids clustering with initial medoids
kmedoids = KMedoids(n_clusters=k, random_state=0, method='pam').fit(data)
labels = kmedoids.labels_
medoids = kmedoids.cluster_centers_

# Compute the median distance for each cluster to define the radius
radii = []
for i, medoid in enumerate(medoids):
    cluster_points = data[labels == i]
    distances = cdist(cluster_points, [medoid], metric='euclidean').flatten()
    median_distance = np.median(distances)
    radii.append(median_distance)

# Plotting the clusters
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', marker='o', label='Data Points')
plt.scatter(medoids[:, 0], medoids[:, 1], c='red', marker='s', s=200, label='Medoids')

# Draw circles around medoids with calculated median radius
for medoid, radius in zip(medoids, radii):
    circle = plt.Circle((medoid[0], medoid[1]), radius, color='red', fill=False, linestyle='dashed')
    plt.gca().add_patch(circle)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('K-Medoids Clustering')
plt.legend()
plt.show()
```
