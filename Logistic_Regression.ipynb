```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score

# Dataset
X = np.array([1, 2, 3, 4, 5, 6])  # Study hours
Y = np.array([0, 0, 0, 1, 1, 1])  # Pass/Fail (Binary)

# Initialize parameters
w0 = 0  # Bias (Intercept)
w1 = 0  # Weight (Slope)
alpha = 0.1  # Learning rate
iterations = 20  # Number of updates to show step-by-step
m = len(X)  # Number of training examples

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Logistic Regression using Gradient Descent
losses = []  # To track the loss over iterations

print(f"{'Iteration':<10}{'w0':<10}{'w1':<10}{'Loss':<15}{'dw0':<10}{'dw1'}")
print("="*55)

for i in range(iterations):
    # Compute prediction (sigmoid activation)
    z = w0 + w1 * X
    h = sigmoid(z)

    # Compute Loss (Binary Cross-Entropy)
    loss = -np.mean(Y * np.log(h) + (1 - Y) * np.log(1 - h))
    losses.append(loss)

    # Compute Gradients
    dw0 = np.mean(h - Y)
    dw1 = np.mean((h - Y) * X)

    # Print step-by-step updates
    print(f"{i+1:<10}{w0:.4f}   {w1:.4f}   {loss:.6f}   {dw0:.4f}   {dw1:.4f}")

    # Update Weights
    w0 -= alpha * dw0
    w1 -= alpha * dw1

# Print final parameters
print("\nFinal Weights:")
print(f"w0: {w0:.4f}, w1: {w1:.4f}")

# Predictions (Threshold at 0.5)
predictions = (sigmoid(w0 + w1 * X) >= 0.5).astype(int)

# Evaluate Accuracy
accuracy = accuracy_score(Y, predictions)
print(f"Accuracy: {accuracy * 100:.2f}%")

# Plot Loss Curve
plt.plot(range(iterations), losses, marker='o', linestyle='-')
plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Loss Curve during Training")
plt.grid()
plt.show()

# Plot Decision Boundary
plt.scatter(X, Y, c=Y, cmap="bwr", edgecolors="k", label="Actual Data")
x_values = np.linspace(0, 7, 100)
y_values = sigmoid(w0 + w1 * x_values)
plt.plot(x_values, y_values, label="Logistic Regression Model", color='black')

plt.axhline(0.5, color='gray', linestyle="--", label="Decision Boundary")
plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.legend()
plt.show()
```
