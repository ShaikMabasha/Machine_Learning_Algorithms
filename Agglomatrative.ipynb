```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

# Step 1: Define Sample Dataset (2D points)
X = np.array([
    [1,2],  # A
    [2,3],  # B
    [5,6],  # C
    [8,8],  # D
    [9,10],  # E
])

# Step 2: Plot Dendrogram
plt.figure(figsize=(8, 5))
linkage_matrix = sch.linkage(X, method='average')  # Using average linkage
sch.dendrogram(linkage_matrix,labels=['A','B','C','D','E'])
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()


# Step 3: Apply Agglomerative Clustering
n_clusters = 3  # Define number of clusters
agg_clustering = AgglomerativeClustering(n_clusters=n_clusters, linkage='average')
labels = agg_clustering.fit_predict(X)

# Print cluster labels
print("Cluster Labels:", labels)

# Step 4: Visualize Clusters
plt.figure(figsize=(8, 5))
for i in range(n_clusters):
    plt.scatter(X[labels == i, 0], X[labels == i, 1], label=f"Cluster {i+1}")

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.legend()
plt.title("Agglomerative Clustering Result")
plt.show()
```

```python
import numpy as np
data = np.array([
    [1,2],  # A
    [2,3],  # B
    [5,6],  # C
    [8,8],  # D
    [9,10],  # E
])

print(data)
```

```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=3,linkage='average')
agg.fit(data)
labels = agg.labels_
print(labels)
```

```python
import matplotlib.pyplot as plt
for i in range(3):
    plt.scatter(data[labels==i,0],data[labels==i,1],label=f'Cluster {i+1}')

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("Agglomerative Clustering Result")
plt.legend()
plt.show()
```

```python
#Dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage
plt.figure(figsize=(8,5))
link_mat = linkage(data,method = 'average')
dendrogram(link_mat,labels=['A','B','C','D','E'])
plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

```python
import numpy as np
data = np.array([[1,2],[2,3],[5,6],[8,8],[9,10]])

#Model
from sklearn.cluster import AgglomerativeClustering
clusters = 3
agg = AgglomerativeClustering(n_clusters=clusters,linkage='average')
agg.fit(data)
labels = agg.labels_
print(labels)

#Plotting
import matplotlib.pyplot as plt
for i in range(clusters):
  plt.scatter(data[labels==i,0],data[labels==i,1],label=f'Cluster {i+1}')
plt.legend()
plt.show()

#Dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage
plt.figure(figsize=(8,5))
link_mat = linkage(data,method = 'single')
dendrogram(link_mat,labels=['A','B','C','D','E'])
plt.show()

#1,10,12,15,17,18,21,22,23,27,30,31,32,33,35,36,40,41,43,48,49,52,56,58,59,61,62,L-1,2,
```
